Default behavior is to act on a .tex file (named below) in your top-level Google Drive, and produce a .pdf file there as well.

You will need to allow Colab access to your Google Drive.

```python
##### Change this line to match the file you want to compile.
filename="Copy of v5a_11a_J10p1_Intro_Harm_Motion_2026June22_Flash35"
#####
tex_filename = filename+".tex"
```

```python
import os
import subprocess
import shutil
import time # for timing the process
_notebook_start = time.time()
```

```python
print("=" * 70)
print("STEP 1: Mount Drive & Download Custom TeX Tarball")
print("=" * 70)

from google.colab import drive
drive.mount('/content/drive')

!mkdir -p /tmp/mini-lualatex
%cd /tmp/mini-lualatex

# Replace with your actual release asset URL once uploaded
# Should work with version currently on github, you just might need to change the version.
TARBALL_URL = "https://github.com/cduston44/mini-lualatex/releases/download/v0.0.3/texlive-physics-tagged.tar.gz"

!wget -q {TARBALL_URL} -O mini-lualatex.tar.gz
!tar -xzf mini-lualatex.tar.gz
```

```python
print("\n" + "=" * 70)
print("STEP 2: Set PATH to Custom Install")
print("=" * 70)

# We'll need to confirm this path once we see the extracted structure
bin_dir = "/tmp/mini-lualatex/bin/x86_64-linux"
os.environ["PATH"] = f"{bin_dir}:{os.environ['PATH']}"

!which lualatex
!lualatex --version
```

```python
print("\n" + "=" * 70)
print("STEP 3: Compile .tex File from Drive")
print("=" * 70)

tex_dir = "/content/drive/MyDrive"  # top-level Drive folder

result = subprocess.run(
    ["lualatex", "--interaction=nonstopmode", "--halt-on-error", tex_filename],
    cwd=tex_dir, capture_output=True, text=True
)

if result.returncode == 0:
    print("SUCCESS: Compiled cleanly.")
else:
    print("FAILED: Compilation errors:")
    print(result.stdout[-1500:])
```

Now create the file in your drive and also trigger browser download.
